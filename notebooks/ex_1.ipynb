```python
# Imports and dependencies
import sys
import os
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), '../')))

import differentialgeometry as dg
import sympy as sp
```

```python
# Creating a euclidean 3d metric - the identity metric 
euclidean_metric = sp.Matrix([  [1, 0, 0],
                                [0, 1, 0],
                                [0, 0, 1]])

x1, x2, x3 = sp.symbols('x1 x2 x3') #three dimensions requires three coordinates

space = dg.Space(metric=euclidean_metric,
                 coord_vars=sp.Matrix([x1, x2, x3]))
```

Now you've created a space. This is the most fundamental operator in the differential geometry libary. From only giving a metric and the variables in the space, one is able to initialise lots of information regarding the space.

Furthermore, any geometrical which you want to create such as a surface, or a line, and operations on these objects, must exist within a space. 

We move on to the first topic of the week, which is about mappings and diffeomorphisms.

```python
# Exercise 1.13

"""
We need a map from R2 to the same space R2. Thus, we construct a euclidean 2d-space.
"""
euclid2d = dg.Space(metric=sp.Matrix([[1, 0],
                                     [0, 1]]),
                    coord_vars=sp.Matrix([x1, x2]))
"""
The domain space and the codomain space is the euclid2d space just constructed. 
The mapping is given by the equation in 1.13.
"""
phi = dg.Map(expr=sp.Matrix([sp.exp(x1) + x2, x1]),
             domain_space=euclid2d,
             codomain_space=euclid2d)

# a) show that phi is a diffeomorphism
print(phi.is_diffeomorphism)

# b) Find Jphi and Jphi^-1
print("Jphi = ", phi.j)

finv, Y = phi.inverseMapping()
phiInv = dg.Map(finv, Y)
print("Jphi^-1 = ", phiInv.j)

# c) Show that these jacobian matrices are the inverses of each other
print("J^-1 phi = ", phi.j.inv())
print("JphiInv^in X", phiInv.j.subs({Y[0]: phi.expr[0], Y[1]: phi.expr[1]}))

if phi.j.inv() == phiInv.j.subs({Y[0]: phi.expr[0], Y[1]: phi.expr[1]}):
    print("The jacobians are inverses of each other")
```
