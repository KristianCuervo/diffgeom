## 1.1 Open Sets (and measuring distances in spaces)
This course considers 2d open sets $\mathcal{U}^2$ (among other higher dimensional sets). To indicate that this is a normed set where a dot product is inherited from the norm, we denote the space as: $(\mathbb{R}^2, \cdot)$. The dot product refers to the **Euclidean distance** between two points in \mathbb{R}^2. 

where the dot is used to make clear that initially we apply the usual dot-product in $\mathbb{R}^2$ and thence the usual **Euclidean distance** between points $p$ and $q$ in $\mathbb{R}^2$:  
If $p$ has coordinates $p = (p^1, p^2)$ and $q = (q^1, q^2)$ then

\[
d_E(p, q) = \sqrt{(q^1 - p^1)^2 + (q^2 - p^2)^2}.
\tag{1.1}
\]
